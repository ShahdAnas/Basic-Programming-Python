```python
# 1
password=(input("Enter Password"))
count=0
has_digit=False
has_uppercase=False

for i in password:
    count+=1
    if i.isdigit():
        has_digit=True
    if i.isupper():
        has_uppercase=True
if count>=8 and has_digit and has_uppercase:
    print ("valid password")
else:
    print("invalid password")
```

```python
# 2
sentence=input()
vowels=0
consonants=0
digits=0
spaces=0
for i in sentence:
    if i in "aeiou":
        vowels+=1
    elif i.isalpha():
        consonants+=1
    elif i.isdigit():
        digits+=1
    elif i.isspace():
        spaces+=1
print (f"vowels:{vowels}  \nconsonants:{consonants} \ndigits:{digits} \nspaces:{spaces}")
```

```python
# 3
grades = [75, 48, 90, 62, 35, 88, 51]
passed=0
failed=0

for i in grades:
    if i >=50:
        passed+=1
    elif i<50:
        failed+=1
print(f"{passed} students passed the exam")
print(f"{failed} students failed the exam")
```

```python
# 4
numbers = [15, 42, 8, 91, 23, 4, 55]
largest = numbers[0]
smallest = numbers[0]
for i in numbers:
    if i > largest:
        largest=i
    elif i < smallest:
        smallest=i
print (f"largest number = {largest}")
print (f"smallest number = {smallest}")
```

```python
# 6
numbers = [5, 2, 5, 8, 2, 10, 8, 12]
unique=[]
for i in numbers:
    if i not in unique:
        unique.append(i)
print(unique)
```

```python
# 7
words = ["apple", "banana", "avocado", "blueberry", "cherry", "apricot"]
dictionary2={}
for i in words:
    f_letter=i[0]
    if f_letter not in dictionary2:
        dictionary2[f_letter]=[]
    dictionary2[f_letter].append(i)

print (dictionary2)
```
